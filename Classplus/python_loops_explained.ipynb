# Loops in Python

Loops are used to **repeat a block of code** as long as a certain condition is met or for a fixed number of iterations.
Python provides two main types of loops:

1. **for loop** – iterate over a sequence (like a list, tuple, string, or range).
2. **while loop** – repeat code while a condition remains True.

This notebook explains their definitions, working, common methods/functions, and examples with comments.


## 1️⃣ For Loop

### Definition
A `for` loop is used to **iterate over a sequence** (such as a list, tuple, dictionary, set, or string) and execute a block of code for each item.

### How It Works
- The loop takes each element of the sequence one by one.
- Executes the code block for that element.
- Stops when there are no more items.

### Syntax
```python
for variable in sequence:
    # code block
```

### Useful Functions with `for`
- **`range(start, stop, step)`**: Generate a sequence of numbers.
- **`enumerate(sequence)`**: Get index and value while looping.
- **`zip(seq1, seq2)`**: Loop over multiple sequences in parallel.





````python
# Example 1: Loop through a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)  # prints each fruit
````

````python
# Example 2: Using range() to loop through numbers
for i in range(10):
    print("Number:", i)
````

## 2️⃣ While Loop

### Definition
A `while` loop **executes a block of code as long as the condition is True**.

### How It Works
- Check the condition.
- If True, execute the block.
- Repeat until the condition becomes False.

### Syntax
```python
while condition:
    # code block
```

````python

# Example 1: Count from 1 to 5
count = 1
while count <= 5:
    print("Count:", count)
    count += 1
````

## 🔁 Loop Control Statements

Python provides statements to change the normal flow of loops:
- **break**: Exit the loop immediately.
- **continue**: Skip the current iteration and move to the next.
- **pass**: Do nothing (placeholder).



````python
### Examples

# break example
for num in range(10):
    if num == 5:
        break  # stop the loop when num is 5
    print(num)
````

````python
# continue example
for num in range(10):
    if num == 5:
        continue  # skip when num is 2
    print(num)
````

````python
# pass example
for num in range(3):
    if num == 1:
        pass  # placeholder, does nothing
    print("Number:", num)
````
