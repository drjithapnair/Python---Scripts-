
# Python Data Structures: Dictionaries

This notebook covers Python **dictionaries**, including:
- Overview
- Dictionary creation
- Accessing elements
- Operations
- Updating and adding elements
- Deletions



## 1. Overview of Dictionaries

A **dictionary** in Python is:
- Unordered (insertion order is preserved from Python 3.7+)
- Mutable (can be changed)
- Stores **key-value pairs**
- Keys must be unique and immutable (string, number, tuple)


```python

# Example: Dictionary Overview
student = {"name": "Alice", "age": 21, "grade": "A"}
print("Dictionary:", student)
print("Type:", type(student))
```


## 2. Dictionary Creation

You can create dictionaries in multiple ways:
- Using curly braces `{}`
- Using the `dict()` constructor


```python

# Different ways to create dictionaries
dict1 = {"brand": "Toyota", "model": "Corolla", "year": 2020}
dict2 = dict(name="John", age=30, city="New York")

print(dict1)
print(dict2)
```


## 3. Accessing Elements

Access values using their keys with `[]` or `get()`.


```python

student = {"name": "Alice", "age": 21, "grade": "A"}

print("Name:", student["name"])
print("Age:", student.get("age"))
print("Grade:", student.get("grade"))
```


## 4. Dictionary Operations

Some common operations include:
- Checking membership (in / not in)
- Iterating over keys, values, or items
- Getting length of dictionary


```python

student = {"name": "Alice", "age": 21, "grade": "A"}

# Membership test
print("Is 'name' a key?", "name" in student)
print("Is 'salary' a key?", "salary" not in student)

# Iteration
print("Keys:")
for key in student.keys():
    print(key)

print("Values:")
for value in student.values():
    print(value)

print("Items (key-value pairs):")
for key, value in student.items():
    print(key, ":", value)

# Length
print("Number of items:", len(student))
```


## 5. Updating and Adding Elements

You can add new key-value pairs or update existing ones.


```python

student = {"name": "Alice", "age": 21}

# Add new key-value pair
student["grade"] = "A"
print("After adding grade:", student)

# Update existing key
student["age"] = 22
print("After updating age:", student)

# Update multiple keys at once
student.update({"city": "New York", "grade": "A+"})
print("After update method:", student)
```


## 6. Deletions

There are several ways to delete items from a dictionary:
- `pop(key)` → removes a key and returns its value
- `popitem()` → removes the last inserted key-value pair
- `del` → removes a key or the entire dictionary
- `clear()` → empties the dictionary


```python

student = {"name": "Alice", "age": 21, "grade": "A", "city": "New York"}

# pop()
removed = student.pop("grade")
print("After pop('grade'):", student, "| Removed:", removed)

# popitem()
last_item = student.popitem()
print("After popitem():", student, "| Removed:", last_item)

# del (specific key)
del student["age"]
print("After del student['age']:", student)

# clear()
student.clear()
print("After clear():", student)
```
