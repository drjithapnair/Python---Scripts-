
# Python Basics: Variables, Data Types, and Operators

This notebook covers the basics of Python programming, including:
- Variables
- Data Types
- Operators

Each section comes with explanations and examples.



## 1. Variables

Variables are used to store data in Python.  
You don't need to declare the type explicitly, Python infers it automatically.


```python

# Example: Variables
x = 10        # integer
y = 3.14      # float
name = "John" # string
is_active = True # boolean

print("x =", x)
print("y =", y)
print("name =", name)
print("is_active =", is_active)
```


## 2. Data Types

Python has several built-in data types:
- int (Integer)
- float (Floating-point number)
- str (String)
- bool (Boolean)



```python

# Examples of different data types
a = 42               # int
b = 3.14159          # float
c = "Hello, Python!" # string
d = False            # boolean


print(type(a), a)
print(type(b), b)
print(type(c), c)
print(type(d), d)

```


## 3. Operators

Operators are used to perform operations on variables and values.

### Types of Operators:
1. Arithmetic Operators: `+`, `-`, `*`, `/`, `//`, `%`, `**`
2. Comparison Operators: `==`, `!=`, `>`, `<`, `>=`, `<=`
3. Logical Operators: `and`, `or`, `not`
4. Assignment Operators: `=`, `+=`, `-=`, `*=`, `/=`, etc.


```python

# Arithmetic Operators
x = 10
y = 3
print("x + y =", x + y)
print("x - y =", x - y)
print("x * y =", x * y)
print("x / y =", x / y)
print("x // y =", x // y)
print("x % y =", x % y)
print("x ** y =", x ** y)

# Comparison Operators
print("x == y:", x == y)
print("x != y:", x != y)
print("x > y:", x > y)
print("x < y:", x < y)
print("x >= y:", x >= y)
print("x <= y:", x <= y)

# Logical Operators
a, b = True, False
print("a and b:", a and b)
print("a or b:", a or b)
print("not a:", not a)

# Assignment Operators
z = 5
z += 2
print("z after += 2:", z)
z *= 3
print("z after *= 3:", z)
```
