# Set in Python

A **set** is a built-in Python data structure that:
- Stores **unordered**, **mutable**, and **unique** elements.
- Does **not allow duplicates**.
- Supports mathematical set operations like **union, intersection, difference**.


## Creating a Set

```python
# Creating a set
fruits = {"apple", "banana", "cherry"}
print(fruits)

# Using set() constructor (duplicates removed automatically)
numbers = set([1, 2, 3, 4, 4, 5])
print(numbers)
```

## Properties of a Set

- **Unordered** → No indexing or slicing.  
- **Unique elements only**.  
- **Mutable** → You can add/remove items.  
- **Heterogeneous elements allowed** (mix of types).  


## Basic Operations on Sets

```python
# Initialize sets
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Union
print(A | B)   
print(A.union(B))

# Intersection
print(A & B)   
print(A.intersection(B))

# Difference
print(A - B)   
print(A.difference(B))

# Symmetric Difference
print(A ^ B)   
print(A.symmetric_difference(B))
```

## Adding and Removing Elements

```python
myset = {1, 2, 3}

# Add single element
myset.add(4)

# Add multiple elements
myset.update([5, 6])

# Remove element (KeyError if not found)
myset.remove(2)

# Discard element (no error if not found)
myset.discard(10)

# Pop element (removes random item)
removed = myset.pop()

# Clear all elements
myset.clear()

print(myset)
```

## Set Functions

```python
nums = {1, 2, 3, 4, 5}

print(len(nums))        
print(min(nums))        
print(max(nums))        
print(sum(nums))        
print(sorted(nums))     
```
