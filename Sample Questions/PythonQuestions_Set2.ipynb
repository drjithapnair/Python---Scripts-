### Check Armstrong Number

```python
num = int(input())
order = len(str(num))
sum = 0
temp = num
while temp > 0:
    digit = temp % 10
    sum += digit ** order
    temp //= 10
print("Armstrong" if sum == num else "Not Armstrong")
```

### Recursive Function for Factorial

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))
```
